# Copying sample_data to Unity Catalog Landing_path

```python
your_sample_data_path = "/Workspace/Repos/adb-emily@startsteps.org/vattenfall-week9-capstone-EmilyImunde/sample_data"

landing_base_path = "/Volumes/vattenfall_dev/raw/landing"

domains = ["market_prices", "weather", "grid_events", "reference"]

for domain in domains:
    source_path = f"{your_sample_data_path}/{domain}"
    target_path = f"{landing_base_path}/{domain}"

    dbutils.fs.mkdirs(target_path)

    print(f"Copying {domain} files to landing volume")

    files = dbutils.fs.ls(source_path)

    for file in files:
        if file.name.endswith(".csv"):
            dbutils.fs.cp(file.path, f"{target_path}/{file.name}", True)
            print("Copied:", file.name)
```
