# Bronze validation summary for all 4 tables

```python
from pyspark.sql import Row

bronze_tables = [
    "vattenfall_dev.raw.bronze_market_prices",
    "vattenfall_dev.raw.bronze_weather",
    "vattenfall_dev.raw.bronze_grid_events",
    "vattenfall_dev.raw.bronze_asset_reference",
]

validation_rows = []

for table_name in bronze_tables:
    df = spark.table(table_name)
    validation_rows.append(
        Row(
            table_name=table_name,
            row_count=df.count(),
            column_count=len(df.columns),
            has_ingestion_ts="ingestion_ts" in df.columns,
            has_source_file="source_file" in df.columns,
        )
    )

validation_summary_df = spark.createDataFrame(validation_rows)

display(validation_summary_df)
```
